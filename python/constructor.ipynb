Constructor

```python
class College:
    college_name = None # class variable
    def __init__(self,name,year,college="bkbiet"):
        self.name = name  # these are instance variables 
        self.year = year
        self.college_name = college
    def show(self):
        print(self.name)
        print(self.college_name)
        print(self.year)
st = College("Himanshu",3)
st1 = College("Mohit",2,"bits")
```

```python
st.show()
st1.show()
```

```python
class Company:
    def __init__(self):
        print("1")

class Client(Company):
    def __init__(self):
        super().__init__()
        print("2")
c = Client()
```
