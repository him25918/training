```python
#  inheritance
# ### Single inheritance

class Company:
    pass

class Developer(Company):   # developer inherited the class company
    pass
```

```python
class Company:
    com_name = "google"
    com_add = "delhi"
class Developer(Company):
    dev_name = "Himanshu"
d = Developer()
```

```python
# multi level inheritance
class Company:
    com_name = "abc"
class Developer(Company):
    dev_name = "himanshu"
class Client(Developer):
    cleint = "xyz"
cl = Client()
print(cl.com_name)
print(cl.dev_name)
print(cl.cleint)
```

```python
# multiple inheritance

class Company:
    pass

class Client:
    pass

class Devloper(Company,Client):
    pass
```

```python
# Hierarchial inheritance
class Company:
    pass

class Developer(Company):
    pass

class Mangers(Company):
    pass
```
