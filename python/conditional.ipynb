```python
student = {
    "roll_no": 101,
    "name": "Himanshu Sharma",
    "age": 19,
    "gender": "Male",
    "course": "B.Tech CSE",
    "semester": 4,
    "section": "A",
    "college": "ABC Institute of Technology",
    "email": "himanshu@example.com",
    "phone": "9876543210"
}
key = input("Enter the key : ")
flag = False
for i in student.keys():
    if(i==key):
        print("key found")
        print(student[i])
        flag = True
        break
if(flag == False):
    print("key not found")
```
