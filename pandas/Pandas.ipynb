```python
import pandas as pd
import numpy as np
# Data preprocessing

# Read data
data = pd.read_csv("Mall_Customers.csv")    # reading the main csv
data.to_csv("Result.csv")        # copying the csv
df = pd.DataFrame(data)          # data ko csv se df

print(df.info)
```

```python
df.isnull() # checking null values
```

```python
df.isnull().sum()
```

```python
df.loc[0,"Age"] = np.nan    # creating null value
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)  # handling the null value
df.to_csv("Result.csv")
```
