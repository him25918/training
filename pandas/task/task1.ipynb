```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("Titanic_test.csv")

df = data.copy()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['Fare'].fillna(df['Fare'].mean(),inplace=True)
df['Cabin'].fillna(df['Cabin'].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['Sex'].unique()
df['Sex'] = df['Sex'].map({'male':0,'female':1})
```
