```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Clean.csv")
```

```python
group2 = df.groupby("Pclass")["Fare"].mean()
group2.plot(kind="line")
```

```python
group2
```

```python
plt.scatter(df["Age"],df["Fare"],color="magenta")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Scatter plot")
```

```python
gender_ratio = df["Sex"].value_counts()
gender_ratio
```

```python
plt.pie(gender_ratio, labels = ["Male","Female"], autopct="%1.1f%%")
```

```python
plt.boxplot(df["Fare"])       # find outliers
```

```python
plt.fill_between(
    group2.index,
    group2.values,
    color="cyan"
)
```

```python
fig, ax = plt.subplots(2,1)
ax[0] = ax[0].scatter(df["Age"],df["Fare"],color="magenta")
ax[1] = ax[1].bar(df["Age"],df["Fare"],color="cyan")
```
